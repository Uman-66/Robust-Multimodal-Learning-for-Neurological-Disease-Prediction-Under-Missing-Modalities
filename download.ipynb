```python
import pandas as pd
import numpy as np
import os
from sklearn.model_selection import train_test_split

# --- UPDATE THESE PATHS TO MATCH YOUR FOLDER ---
CSV_PATH = "oasis_cross-sectional.csv"  # Path to your CSV
IMAGES_DIR = "oasis_images/oasis_preprocessed"             # Path to your folder containing the .npy files

# 1. Load the clinical CSV
df = pd.read_csv(CSV_PATH)
print(f"Total subjects in CSV: {len(df)}")
print(df.head())
print(df.columns)
```

```python
# 2. Create the Target (Binary Classification: Healthy vs. Impaired)
df['target'] = (df['CDR'] > 0).astype(int)
print(f"Class distribution:\n{df['target'].value_counts()}")

# 3. Prepare Clinical Modality (Modality 2)
# Encode Sex: M -> 0, F -> 1
df['Sex_encoded'] = df['M/F'].map({'M': 0, 'F': 1})

# Select the columns for your clinical vector
clinical_columns = ['MMSE', 'Age', 'Educ', 'Sex_encoded']
# Handle any missing values (drop rows where clinical data is NaN)
df = df.dropna(subset=clinical_columns + ['target'])
print(f"Subjects after dropping missing clinical data: {len(df)}")

# 4. Create the MRI file paths (Modality 1)
# The CSV has an 'ID' column. The .npy files are named {ID}_t1_masked.npy
df['mri_path'] = df['ID'].apply(lambda x: os.path.join(IMAGES_DIR, f"{x}_t1_masked.npy"))

# Check if the files actually exist on disk
df['file_exists'] = df['mri_path'].apply(os.path.exists)
print(f"Subjects with MRI files found: {df['file_exists'].sum()}")
# Keep only the rows where the MRI file exists
df = df[df['file_exists']].reset_index(drop=True)
print(f"Final dataset size: {len(df)}")
```

```python
# 5. Split the data
# First, separate 20% for final testing
train_val_df, test_df = train_test_split(df, test_size=0.2, random_state=42, stratify=df['target'])

# Then, split the training data into 80% train and 20% validation (which is 80% of 80% = 64% of total, 16% val)
train_df, val_df = train_test_split(train_val_df, test_size=0.2, random_state=42, stratify=train_val_df['target'])

print(f"Train size: {len(train_df)}, Val size: {len(val_df)}, Test size: {len(test_df)}")
```

```python
import torch
from torch.utils.data import Dataset

class OASISMultimodalDataset(Dataset):
    def __init__(self, dataframe):
        self.df = dataframe.reset_index(drop=True)
        
    def __len__(self):
        return len(self.df)
    
    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        
        # --- Modality 1: Imaging (3D MRI) ---
        # Load the .npy file. Shape is (96, 128, 96)
        mri_volume = np.load(row['mri_path']).astype(np.float32)
        # Add a channel dimension for the CNN: (1, 96, 128, 96)
        mri_volume = np.expand_dims(mri_volume, axis=0)
        
        # --- Modality 2: Clinical (Tabular) ---
        # Extract the clinical values
        clinical_data = np.array([
            row['MMSE'],
            row['Age'],
            row['Educ'],
            row['Sex_encoded']
        ], dtype=np.float32)
        
        # --- Target ---
        label = np.array(row['target'], dtype=np.int64)
        
        # Convert to PyTorch tensors
        mri_tensor = torch.tensor(mri_volume, dtype=torch.float32)
        clinical_tensor = torch.tensor(clinical_data, dtype=torch.float32)
        label_tensor = torch.tensor(label, dtype=torch.long)
        
        return mri_tensor, clinical_tensor, label_tensor
```

```python
from torch.utils.data import DataLoader

# Create the datasets
train_dataset = OASISMultimodalDataset(train_df)
val_dataset = OASISMultimodalDataset(val_df)
test_dataset = OASISMultimodalDataset(test_df)

# Create DataLoaders (Batch size depends on your GPU memory. Start with 4 or 8)
BATCH_SIZE = 4  # Increase if you have a strong GPU

train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
test_loader = DataLoader(test_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

print("DataLoaders are ready!")
print(f"Batches in Train: {len(train_loader)}")
```
